# Inspection, adoption and disclosure

This notebook shows the objects of the last three cases of the acceptance of the Matsya chat integration in the Workbench: a colleague who opens the author's folder under another account, a job's returned files read as a model of their own and adopted into the model only as reviewed, and what each request shows before it is sent. Each cell is one step that the acceptance file `tests/test_spec_0_6_unit_g5.py` also takes, against the stand-in of the client or a Matsya service started in this process from the service package's test configuration. No step reaches the installed service.

The first cell puts the notebook in this checkout, makes a scratch folder for every copy and makes the stand-in the client the Workbench runs; `matsya_cases.py` beside this notebook is the one helper the reader need not read.

The durables scratch copies omit methods additions because job preparation does not yet accept them. Their stage equations and fixed calibration profiles are copied from the application. The notebook checks declaration exchange; it does not solve the economic model.

```python
import asyncio
import matsya_cases
from bellman_workbench import matsya_client, matsya_window, matsya_work, recipe, returned, trellis
```

## 4 A replacement over unsaved text

The calibration file holds unsaved text in the editor.

```python
buffers = matsya_cases.unsaved(folder, matsya_cases.CALIBRATION, "# typed, not saved\n")
```

The plan of `Replace model files…` names the reviewed version, the destination and the line differences, the editor's included, and refuses the adoption.

```python
returned.plan(folder, copy, buffers=buffers, changed_only=True)
```

The adoption itself checks the editor again immediately before writing, refuses and writes nothing.

```python
before = matsya_cases.fingerprint(folder)
```

```python
returned.adopt(folder, returned.plan(folder, copy, changed_only=True), buffers=buffers)
```

```python
matsya_cases.fingerprint(folder) == before
```

## 6 A late write failure recorded and recovered

Iteration 3 changes the calibration and the settings file. Its adoption is written through a writer that fails on the second file with the operating system's full-disk error.

```python
two = matsya_cases.lowered_with_settings(folder, session, number, 3)
```

```python
originals = matsya_cases.files_of(folder)
```

```python
stopped = returned.adopt(folder, returned.plan(folder, two, changed_only=True), write=matsya_cases.full_disk)
```

The adoption stops with its sentence; its record names the file written and the one that failed, each original kept in the adoption's recovery area.

```python
stopped.refusals
```

```python
stopped.record["files"]
```

The restore writes the originals back and is recorded.

```python
returned.restore(folder, stopped.id).record["state"]
```

The model files are again those before the adoption; only the adoption's own record under `.bellman-matsya/` was added.

```python
matsya_cases.changed_since(folder, originals)
```

## 8 What each request shows before it is sent

Matsya's column for a fourth copy, drawn in this process with the page's own components. Calls that wait for its background tasks run with `asyncio.to_thread`, so the notebook's event loop remains available to those tasks.

```python
column = matsya_cases.Column(shown)
```
