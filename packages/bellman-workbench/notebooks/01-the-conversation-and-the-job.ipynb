# The conversation and the job

This notebook shows the objects of the first three cases of the acceptance of the Matsya chat integration in the Workbench: what the Workbench prepares from a model folder and sends to Matsya, what the service reads of a session's history, how a reply continues a job, and how accepted work is found again after the page closes. Each cell is one step that the acceptance file `tests/test_spec_0_6_unit_g5.py` also takes. Two services answer: the stand-in of the client, a small program of the Workbench's tests that keeps its service in a file, and a Matsya service started in this process from the service package's test configuration, whose jobs end with one question and which calls no model provider. No step reaches the installed service.

The first cell puts the notebook in this checkout, makes a scratch folder for every copy and makes the stand-in the client the Workbench runs; `matsya_cases.py` beside this notebook is the one helper the reader need not read.

The durables scratch copies omit methods additions because job preparation does not yet accept them. Their stage equations and fixed calibration profiles are copied from the application. The notebook checks declaration exchange; it does not solve the economic model.

```python
import matsya_cases
from bellman_workbench import matsya_client, matsya_window, matsya_work, recipe, returned, trellis
```

## 1 The pension model's files prepared for a build

A scratch copy of the pension model, whose trellis file reads two period templates: the working years read `period.working.yml` and the other years `period.pensions.yml`. These are synthetic files for checking Workbench operations; their calibration does not reproduce the paper's pension model.

```python
pensions = matsya_cases.pensions()
```

The session of the stand-in's service that the build is sent in: its identifier, and the digest that names the service, computed from the client's settings and never holding the token.

```python
session = matsya_cases.session_of(pensions)
```

```python
session
```

An edit of the shocks stage typed in the editor and not saved: the editor's buffer of `declaration/stages/shocks/shocks.bl`, its text and the digest of the file it was read from.

```python
buffers = matsya_cases.unsaved(pensions, matsya_cases.SHOCKS, "# the edit under review, not yet saved\n")
```

Without a choice for the unsaved text the preparation writes nothing. It answers the sentence and the difference between the disk and the editor, which the column shows with the three choices `Save`, `Saved version` and `Displayed text`.

```python
matsya_work.prepare(pensions, matsya_cases.PENSIONS, "recipe", session, 1, buffers=buffers)
```

With the choice `displayed text` the preparation writes the prepared-inputs file and the Workbench's record beside it, and sends nothing.

```python
prepared = matsya_work.prepare(pensions, matsya_cases.PENSIONS, "recipe", session, 1, buffers=buffers, choice=matsya_work.DISPLAYED_TEXT)
```

The record lists each file the recipe reads: its path in the folder, the name the service reads it under, the fingerprint of the text taken, and whether that text came from the disk or the editor. Both period templates stand, and the shocks stage's text came from the editor.

```python
prepared.record["files"]
```

The recipe is sent as `spec.yml`, the one name the service reads for a recipe, so its `trellis:` entry is rewritten to `trellis.yml`; the record names that one rewriting.

```python
prepared.record["rewritten"]
```

The submission sends the prepared-inputs file as it stands and reads no file of the model again; the stand-in accepts the job.

```python
matsya_work.submit(pensions, prepared.path, session).record
```

The digest of each text the stand-in received, by the name the service reads it under.

```python
matsya_cases.received()
```

They are the digests of the texts of the prepared-inputs file, the texts the user reviewed: the comparison is true.

```python
matsya_cases.received() == matsya_cases.prepared_texts(prepared)
```

## 4 A stale review refused and a reviewed history fixed at acceptance

The Matsya service of the service package's test configuration, started in this process with two principals, the author and a colleague.

```python
service = matsya_cases.Service()
```

```python
service
```

The Workbench runs the client itself against that service, with the author's Matsya token.

```python
matsya_cases.real_client(service.address, matsya_cases.AUTHOR_TOKEN)
```

A second copy of the durables model, connected by `Connect this folder to Matsya`: the client's `project init` creates the folder's project session, enters `economics.md` as its first entry and writes `.bellman-matsya/project.json`.

```python
folder = matsya_cases.durables()
```

```python
matsya_work.connect(folder)
```

```python
project = matsya_work.connected_session(folder)
```

A build's first step enters the request in the session as the user's entry, number 2.

```python
matsya_work.add_entry(folder, project, "Lower the housing cost τ to 0.08.").record["accepted"]
```

A submission prepared when the user had reviewed entry 1 alone is refused: the service names the entry added since and asks for a review. The request's record keeps the sentence, and no job is accepted.

```python
matsya_work.submit(folder, matsya_work.prepare(folder, matsya_cases.DURABLES, "recipe", project, 1).path, project).record
```

Two jobs now hold both of the service's workers, so that the next job waits in the queue.

```python
service.hold()
```

A submission that reviewed entry 2 is accepted.

```python
accepted = matsya_work.submit(folder, matsya_work.prepare(folder, matsya_cases.DURABLES, "recipe", project, 2).path, project)
```

```python
accepted.record["accepted"]
```

While the job waits, the user enters a third entry.

```python
matsya_work.add_entry(folder, project, "Let the wage grow with age.").record["accepted"]
```

Following the job reads its state and writes nothing. It is queued, and the service fixed its history through entry 2 when it accepted it: read `reviewed_through`, `revision` and `entry_numbers`.

```python
matsya_work.follow(accepted.record)
```

```python
service.release()
```

Once the job has ended, its record names the history it read: entries 1 and 2, not entry 3.

```python
matsya_cases.ended(accepted.record)["result"]["captured_source"]
```

The end of the text the service's processor read as the job's source: entry 2 is its last entry.

```python
print(service.processor.sources[0]["text"][-160:])
```

The job's label marks it overtaken, built from an earlier version, since entry 3 stands above the history it read.

```python
matsya_cases.ended(accepted.record)["label"]
```

## 7 A job that did not converge, in five parts

The folder connected to the stand-in's service, and a job of that service that did not converge: its comparison gives one difference, and it carries one comment on the prose and one unresolved assumption.

```python
held = matsya_cases.connected(folder)
```

```python
job = matsya_cases.not_converged(folder, held, 7)
```

`Download returned files` writes one new saved copy of the job's model folder.

```python
copy = matsya_work.download(folder, held, job).path
```

```python
copy
```

The provisional calibration the job returned is kept in the copy: it reads τ 0.08, while the folder's own file still reads 0.12.

```python
print(open(folder.root / copy / matsya_cases.CALIBRATION).read())
```

## 8 Accepted work found again after the page closes

A third copy of the durables model, connected to the stand-in.

```python
closed = matsya_cases.durables()
```

```python
closing = matsya_cases.connected(closed)
```

A job whose client prints the accepted job and then the warning that the service used its own key, ending with status 1: the record keeps the accepted job, the status and the warning.

```python
warned = matsya_cases.warned_submission(closed, closing)
```

```python
warned.record
```

A turn accepted, whose wait then ends, as when the adapter's limit of 930 seconds passes; its record keeps the turn's identifier.

```python
turn = matsya_work.start_turn(closed, closing, "What does tenure choose?")
```

```python
turn.record["accepted"]
```

A turn whose answer is lost in transport: the service holds the turn, and the record holds no identifier.

```python
lost = matsya_cases.lost_turn(closed, closing, "And what does the keeper choose?")
```

```python
lost.record
```

While the page is closed the service runs the job and answers the first turn.

```python
matsya_cases.service_works_on(warned.record["accepted"], turn.record["accepted"], "Whether to keep the house or to adjust it.")
```

```python
matsya_cases.forget_calls()
```

```python
before = matsya_cases.fingerprint(closed)
```

The adapter's module loaded again, as when the page opens again. Recovery reads each accepted identifier before it offers anything, and offers a retry for the lost turn alone: read `state` and `retry`.

```python
adapter = matsya_cases.reopened()
```

```python
recovered = adapter.recover(closed, closing)
```

```python
recovered
```

Recovery ran `job status` and `turn status` alone and wrote nothing: the folder's fingerprint is the one before.

```python
matsya_cases.calls()
```

```python
matsya_cases.fingerprint(closed) == before
```

The retry sends the lost turn's delivery identifier again, and the service answers the turn it already holds: it holds two turns, not three.

```python
adapter.retry(closed, recovered[2].record).record["accepted"]
```

```python
matsya_cases.stand_in_turns()
```

Following the job and the column's recovery of the folder's jobs read the job's state and list it as a message; neither cancels nor downloads.

```python
matsya_cases.forget_calls()
```

```python
adapter.follow(recovered[0].record)["state"]
```

```python
matsya_window.recover_jobs(closed)
```

```python
matsya_window.exchanges.value
```

```python
matsya_cases.calls()
```

No saved copy stands, since nothing was downloaded.
